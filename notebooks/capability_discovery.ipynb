## Tri-hybrid Capability Search

Run this section after ingesting the workspace and running the capability input/discovery pairs above. The search combines BM25 matches on capability names and summaries, semantic similarity against capability embeddings, and graph-coverage heuristics. The three ranked lists are fused with reciprocal rank fusion (RRF).

````python
user_query = "Which capabilities handle customer account balances and interest?"
ranked_capabilities = discovery_service.discover_capabilities(
    user_query, top_k=5, candidate_k=20, rrf_k=60
)

for rank, capability in enumerate(ranked_capabilities, start=1):
    print(
        f"{rank}. {capability['name']} "
        f"(RRF={capability['rrf_score']:.4f}; "
        f"signals={', '.join(capability['matched_by'])})"
    )
    print(
        f"   Graph coverage: {capability['program_count']} programs, "
        f"{capability['job_count']} jobs, "
        f"{capability['copybook_count']} copybooks"
    )
````
